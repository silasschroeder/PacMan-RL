```python
grid = [
    [0, 0, 0],
    [0, 0, 0],
    [0, 0, 0]
]
```

```python
print(grid[0][1])
```

```python
# 5 points for moving from grid[0][2] to grid[1][2]
# 10 points for moving fromgrid[0][1] to grid[2][1]
# -1 point for each move (up, down, left, right)
# calculate the value function for each cell in the grid
# assuming the agent wants to maximize its score and we have a horizon of 10 moves.
for i in range(3):
    for j in range(3):
        value = 0
        # Move to (1,2)
        if (i, j) == (0, 2):
            value += 5 - 1  # 5 points for reaching (1,2) and -1 for the move
        # Move to (2,1)
        if (i, j) == (0, 1):
            value += 10 - 1  # 10 points for reaching (2,1) and -1 for the move
        grid[i][j] = value
print(grid)
```

```python
#manual calculation:
import numpy as np
one_1 = [-1,10,-1,-1,10,-1,-1,10,-1,-1]
one_2 = [-1,-1,10,-1,-1,10,-1,-1,10,-1]
one_3 = [-1,-1,-1,10,-1,-1,10,-1,-1,10]
two_1 = [10, -1, -1, 10, -1, -1, 10, -1, -1, 10]
two_2 = [-1,10,-1,-1,10,-1,-1,10,-1,-1]
two_3 = [-1,-1,10,-1,-1,10,-1,-1,10,-1]
three_1_a3 = [-1,10,-1,-1,10,-1,-1,10,-1,-1]
three_2_a3 = [-1,-1,10,-1,-1,10,-1,-1,10,-1]
three_3_a3 = [-1,-1,-1,10,-1,-1,10,-1,-1,10]
three_1_b1_a2 = [5,-1,-1,10,-1,-1,10,-1,-1,10]
three_2_b1_a2 = [-1, 5,-1,-1,10,-1,-1,10,-1,-1]
three_3_b1_a2 = [-1,-1,5,-1,-1,10,-1,-1,10,-1]
grid = [
    one_1, one_2, one_3,
    two_1, two_2, two_3,
    three_1_a3, three_2_a3, three_3_a3,
    three_1_b1_a2, three_2_b1_a2, three_3_b1_a2,
]
```

```python
# alpah=0.9
alpha = np.array([0.9 ** i for i in range(10)])
values = [np.dot(np.array(grid[i]),alpha) for i in range(len(grid))]
print(values)
```

```python
# alpah=0.1-0.9
alphas = [0.1 * i for i in range(1,10)]
for a in alphas:
    alpha = np.array([a ** i for i in range(10)])
    values = [round(np.dot(np.array(grid[i]),alpha), 2) for i in range(len(grid))]
    print(values)
```
